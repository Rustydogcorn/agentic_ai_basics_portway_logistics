```python
%sql
create catalog if not exists agentic_ai;
create schema if not exists agentic_ai.portway_logistics;
create volume if not exists agentic_ai.portway_logistics.raw_data;
```

```python
files = dbutils.fs.ls("/Volumes/agentic_ai/portway_logistics/raw_data")

file_valdations = []
for file in files:
    
    df = spark.read.format("csv").option("header","true").option("inferSchema","true").load(file[0])
   
    file_name = file[1].replace(".csv","")
    file_valdations.append(file_name)
   
    df.write.format("delta").mode("overwrite").saveAsTable(f"agentic_ai.portway_logistics.{file_name}")

print("Files written as table:", file_valdations)
```
